**Question 1:** How many customers are in the file?
(as help, we have added some comments and starter code to help you structure the solution)

```python
from os.path import exists
import csv

datafile = "./sample_data/customers.csv"

if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files")

count = 0

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    next(csvreader)

    for row in csvreader:
        count += 1

print(count)
```

**Question 2:** In how many different states do the customers live in?

```python
import csv

states = set()

with open("./sample_data/customers.csv", "r") as fl:
    reader = csv.DictReader(fl)

    for row in reader:
        state = row["State"].strip().upper()  # Normalizar mayúsculas
        if state:
            states.add(state)

print(len(states))
```

**Question 3** What is the state with most customers?

```python
import csv
from collections import Counter

state_count = Counter()

with open("./sample_data/customers.csv", "r") as fl:
    reader = csv.DictReader(fl)

    for row in reader:
        state = row["State"].strip().upper()
        if state:
            state_count[state] += 1

print(state_count.most_common(1))
```

**Question 4** What is the state with the least customers?

```python
min(state_count, key=state_count.get)
```

**Question 5:** What is the most common last name?

```python
import csv

last_name_count = {}

with open("./sample_data/customers.csv", "r") as fl:
    reader = csv.DictReader(fl)

    for row in reader:
        last_name = row["LastName"].strip()

        if last_name in last_name_count:
            last_name_count[last_name] += 1
        else:
            last_name_count[last_name] = 1

most_common = max(last_name_count, key=last_name_count.get)

print(most_common)
```

```python
# Author: Sebastián Guerrero
# Exercise 2: Processing Orders Data

import csv


# Question #1
# Answer: 16672

order_ids = set()

with open('./sample_data/orders.csv', 'r') as f:
    reader = csv.DictReader(f)

    for row in reader:
        order_ids.add(row['OrderID'].strip())

print("Question #1:", len(order_ids))


# Question #2
# Answer: 1.76

total_items = 0
order_ids = set()

with open('./sample_data/orders.csv', 'r') as f:
    reader = csv.DictReader(f)

    for row in reader:
        total_items += 1
        order_ids.add(row['OrderID'].strip())

print("Question #2:", round(total_items / len(order_ids), 2))


# Question #3
# Answer: 35

items_per_order = {}

with open('./sample_data/orders.csv', 'r') as f:
    reader = csv.DictReader(f)

    for row in reader:
        order_id = row['OrderID'].strip()
        items_per_order[order_id] = items_per_order.get(order_id, 0) + 1

print("Question #3:", max(items_per_order.values()))


# Question #4
# Answer: 267

oct_orders = set()

with open('./sample_data/orders.csv', 'r') as f:
    reader = csv.DictReader(f)

    for row in reader:
        if row['Date'].startswith('2021-10'):
            oct_orders.add(row['OrderID'].strip())

print("Question #4:", len(oct_orders))


# Question #5
# Answer:
# Sophie Labonte
# Customer ID: 1909369
# Total spent: $27560.0

customer_totals = {}

with open('./sample_data/orders.csv', 'r') as f:
    reader = csv.DictReader(f)

    for row in reader:
        customer_id = row['CustomerID'].strip()
        order_total = float(row['OrderTotal'])

        customer_totals[customer_id] = customer_totals.get(customer_id, 0) + order_total

top_customer_id = max(customer_totals, key=customer_totals.get)

customer_names = {}

with open('./sample_data/customers.csv', 'r') as f:
    reader = csv.DictReader(f)

    for row in reader:
        customer_names[row['CustomerID'].strip()] = (
            row['FirstName'].strip() + " " + row['LastName'].strip()
        )

print("Question #5:")
print("Name:", customer_names[top_customer_id])
print("Customer ID:", top_customer_id)
print("Total spent:", customer_totals[top_customer_id])


# Question #6
# Answer: 01 (January)

monthly_sales = {}

with open('./sample_data/orders.csv', 'r') as f:
    reader = csv.DictReader(f)

    for row in reader:
        month = row['Date'][5:7]
        total = float(row['OrderTotal'])

        monthly_sales[month] = monthly_sales.get(month, 0) + total

best_month = max(monthly_sales, key=monthly_sales.get)

print("Question #6:", best_month)
```
